```python

# Home Task 2 - Question 1


tree = dict(file_system)

for children in file_system.values():
    for name in children:
        tree.setdefault(name, [])


def dfs_scan(graph, start):
    visited = set()
    stack = [start]
    order = []

    while stack:
        node = stack.pop()

        if node in visited:
            continue

        visited.add(node)
        order.append(node)

        # Reverse to keep the original folder order
        for child in reversed(graph[node]):
            if child not in visited:
                stack.append(child)

    return order


dfs_order = dfs_scan(tree, "/")

print("DFS Traversal Order:")
for i, item in enumerate(dfs_order, 1):
    print(i, item)

print("\nTotal items visited:", len(dfs_order))
```

```python

# Home Task 2 - Question 2


def print_tree(graph, node, depth=0):
    print("    " * depth + node)

    for child in graph[node]:
        print_tree(graph, child, depth + 1)


print_tree(tree, "/")
```

```python

# Home Task 2 - Question 3


def get_full_paths(graph, node, current_path="/"):
    paths = [current_path]

    for child in graph[node]:
        if current_path == "/":
            child_path = "/" + child
        else:
            child_path = current_path + "/" + child

        paths.extend(get_full_paths(graph, child, child_path))

    return paths



all_paths = get_full_paths(tree, "/")

print("Full Paths of All Files and Folders:\n")

for path in all_paths:
    print(path)

print("\nTotal items:", len(all_paths))
```

```python

# Home Task 2 - Question 4


def is_suspicious(name, path):
    size = file_sizes.get(name, 0)

    if name.endswith(".exe"):
        return True

    if name.endswith(".php") and "uploads" in path.split("/"):
        return True

    if "password" in name or "keylogger" in name:
        return True

    if name.startswith(".") and size > 1000:
        return True

    return False


def find_suspicious(graph, node="/", current_path="/"):
    suspicious = []

    
    if node in file_sizes and is_suspicious(node, current_path):
        suspicious.append((current_path, file_sizes[node]))

   
    for child in graph[node]:
        if current_path == "/":
            child_path = "/" + child
        else:
            child_path = current_path + "/" + child

        suspicious.extend(
            find_suspicious(graph, child, child_path)
        )

    return suspicious


suspicious_files = find_suspicious(tree)

print("Suspicious Files Found:\n")

for path, size in suspicious_files:
    print(path, "-", size, "KB")

print("\nTotal suspicious files:", len(suspicious_files))
```

```python

# Home Task 2 - Question 5


def calculate_size(graph, node, folder_sizes):
    
    if node in file_sizes:
        return file_sizes[node]

    total = 0

   
    for child in graph[node]:
        total += calculate_size(graph, child, folder_sizes)

    folder_sizes[node] = total
    return total


folder_sizes = {}


total_size = calculate_size(tree, "/", folder_sizes)


largest_file = max(file_sizes, key=file_sizes.get)

print("Largest File:")
print(largest_file, "-", file_sizes[largest_file], "KB")

print("\nFolder Sizes:")
for folder, size in folder_sizes.items():
    print(folder, "-", size, "KB")

print("\nTotal File System Size:", total_size, "KB")
```

```python

# Home Task 2 - Question 6 (Corrected)

def check_folder(graph, node, current_path, results):
    # If this is a file, check whether it is suspicious
    if node in file_sizes:
        return is_suspicious(node, current_path)

    has_suspicious = False

    # Search all children using DFS
    for child in graph[node]:
        if current_path == "/":
            child_path = "/" + child
        else:
            child_path = current_path + "/" + child

        if check_folder(graph, child, child_path, results):
            has_suspicious = True

    results[current_path] = has_suspicious
    return has_suspicious


folder_results = {}
check_folder(tree, "/", "/", folder_results)

print("Folders WITH suspicious files:\n")
for path, found in folder_results.items():
    if found:
        print(path)

print("\nFolders WITHOUT suspicious files:\n")
for path, found in folder_results.items():
    if not found:
        print(path)

print("\nTotal folders:", len(folder_results))
```

```python

# Home Task 2 - Question 7


from collections import deque

def bfs_suspicious(graph):
    queue = deque([("/", "/")])
    found = []

    while queue:
        node, path = queue.popleft()

        if node in file_sizes and is_suspicious(node, path):
            found.append(path)

        for child in graph[node]:
            if path == "/":
                child_path = "/" + child
            else:
                child_path = path + "/" + child

            queue.append((child, child_path))

    return found



dfs_found = [path for path, size in suspicious_files]


bfs_found = bfs_suspicious(tree)

print("DFS vs BFS Suspicious File Discovery\n")

print("DFS Order:")
for i, path in enumerate(dfs_found, 1):
    print(i, path)

print("\nBFS Order:")
for i, path in enumerate(bfs_found, 1):
    print(i, path)

print("\nFirst file found by DFS:", dfs_found[0])
print("First file found by BFS:", bfs_found[0])
```

```python

# Home Task 2 - Question 8


from collections import deque

def count_depth_levels(graph):
    queue = deque([("/", 0)])
    depth_counts = {}

    while queue:
        node, depth = queue.popleft()

        
        depth_counts[depth] = depth_counts.get(depth, 0) + 1

        
        for child in graph[node]:
            queue.append((child, depth + 1))

    return depth_counts



depth_counts = count_depth_levels(tree)

print("BFS Depth Level Counts:\n")

for depth, count in sorted(depth_counts.items()):
    print("Depth", depth, ":", count, "items")

print("\nTotal items:", sum(depth_counts.values()))
```

```python

# Home Task 2 - Question 9

linked_tree = {
    node: children.copy()
    for node, children in tree.items()
}

linked_tree["alice"].append("shared")
linked_tree["shared"].append("www")
linked_tree["tmp"].append("home")

print("Symbolic links added successfully!")

def dfs_without_visited(graph, start, limit=100):
    stack = [start]
    order = []

    while stack and len(order) < limit:
        node = stack.pop()
        order.append(node)

        for child in reversed(graph[node]):
            stack.append(child)

    return order

without_visited = dfs_without_visited(linked_tree, "/", 100)

print("\nDFS without visited set:")
print("Items processed:", len(without_visited))
print("Repeated items:", len(without_visited) - len(set(without_visited)))

def dfs_with_visited(graph, start):
    stack = [start]
    visited = set()
    order = []

    while stack:
        node = stack.pop()

        if node in visited:
            continue

        visited.add(node)
        order.append(node)

        for child in reversed(graph[node]):
            if child not in visited:
                stack.append(child)

    return order

with_visited = dfs_with_visited(linked_tree, "/")

print("\nDFS with visited set:")
print("Unique items visited:", len(with_visited))
print("Repeated items:", len(with_visited) - len(set(with_visited)))
```

```python

# Home Task 2 - Question 10

from collections import deque

def dfs_path(graph, start, goal):
    visited = set()

    def search(node, path):
        if node in visited:
            return None

        visited.add(node)

        if node == goal:
            return path

        for child in graph[node]:
            result = search(child, path + [child])
            if result is not None:
                return result

        return None

    return search(start, [start])

def bfs_path(graph, start, goal):
    queue = deque([[start]])
    visited = {start}

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        for child in graph[node]:
            if child not in visited:
                visited.add(child)
                queue.append(path + [child])

    return None


dfs_result = dfs_path(linked_tree, "/", "shell.php")
bfs_result = bfs_path(linked_tree, "/", "shell.php")

dfs_www = dfs_path(linked_tree, "/", "www")

print("Total unique items visited:", len(with_visited))

print("\nDFS first reaches /var/www through:")
print(" -> ".join(dfs_www))

print("\nDFS path to shell.php:")
print(" -> ".join(dfs_result))
print("Hops:", len(dfs_result) - 1)

print("\nBFS path to shell.php:")
print(" -> ".join(bfs_result))
print("Hops:", len(bfs_result) - 1)

if len(dfs_result) < len(bfs_result):
    print("\nDFS path is shorter.")
elif len(bfs_result) < len(dfs_result):
    print("\nBFS path is shorter.")
else:
    print("\nBoth paths have equal length.")
```
